#OneHotEncoding

```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder

#sample categorical data
data = np.array([['Red'],['Green'],['Blue'],['Red'],['Blue']])

#initiallize the OneHotEncoder
encoder = OneHotEncoder(sparse_output=False) # sparse_output=False returns an aray

#fit and transform the data
encoded_data = encoder.fit_transform(data)

print(encoded_data)
```

#CountVectorizer is a bag of words

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()

```

```python
bow = cv.fit_transform(df['text'])
```

```python
bow
```

```python
bow.toarray()
```

```python
#vocabulary
print(cv.vocabulary_)
```

#N-Grams

```python
df = pd.DataFrame({
    "text": ["people watch dswithatik",
             "dswithatik watch dswithatik",
             "people write comment",
             "dswithatik write comment"],
    "Output":[1,1,0,0]

                   })
```

```python
# BI grams
cv_bi = CountVectorizer(ngram_range=(2,2)) #try gram (3,3)
```

```python
bow = cv_bi.fit_transform(df['text'])
bow.toarray()
```

```python
#voculabury
print(cv_bi.vocabulary_)
```

#word2vec

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("saurabhbadole/game-of-thrones-book-dataset")

print("Path to dataset files:", path)
```

```python
!pip install gensim
import numpy as np
import pandas as pd
import gensim
import os
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
story = []
for filename in os.listdir(path):
  if filename == '.ipynb_checkpoints':
    pass
  f = open(os.path.join(path, filename), encoding='latin-1')
  corpus = f.read()
  raw_sent = sent_tokenize(corpus)
  for sent in raw_sent:
    story.append(simple_preprocess(sent))
```

```python
story
```

```python
len(story)
```

```python
story[0]
```

```python
model = gensim.models.Word2Vec(
    window=10,
    min_count=2,

)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
vec = model.wv.get_normed_vectors()
```

```python
vec
```
