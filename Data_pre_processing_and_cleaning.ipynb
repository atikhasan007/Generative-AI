#HTML tag remove

```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

```python
df['review'] = df['review'].apply(remove_html_tags)
```

#Remove URL

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

```python
df['review'] = df['review'].apply(remove_url)
```

#Punctuation handling

```python
import string, time
string.punctuation
```

```python
exclude = string.punctuation
exclude
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char,'')
  return text
```

```python
df['review'] = df['review'].apply(remove_punc)
```

```python
df['review'][0]
```

```python
start = time.time()
remove_punc(df['review'][0])
time1 = time.time() - start
print(time1*50)
```

#More optimize function for punctuation remove

```python
def remove_punc1(text):
  return text.translate(str.maketrans('', '', exclude))
```

```python
start = time.time()
remove_punc1(df['review'][0])
time1 = time.time() - start
print(time1*50)
```

#Chat_conversion handle

```python
cha_words = {
    "AFAIK":"As Far As I Know",
    "AFK":"Away From Keyboard",
    "ASAP":"As Soon As Possible",
    "ATK":"At The Keyboard",
    "ATM":"At The Moment",
    "A3":"Anytime, Anywhere, Anyplace",
    "BAK":"Back At Keyboard",
    "BBL":"Be Back Later",
    "BBS":"Be Back Soon",
    "BFN":"Bye For Now",
    "B4N":"Bye For Now",
    "BRB":"Be Right Back",
    "BRT":"Be Right There",
    "BTW":"By The Way",
    "B4":"Before",
    "B4N":"Bye For Now",
    "CU":"See You",
    "CUL8R":"See You Later",
    "CYA":"See You",
    "FAQ":"Frequently Asked Questions",
    "FC":"Fingers Crossed",
}
```

```python
def chat_conversion(text):
  new_text = []
  for w in text.split():
    if w.upper() in cha_words:
      new_text.append(cha_words[w.upper()])
    else:
      new_text.append(w)
  return " ".join(new_text)
```

#Stopwords

```python
from nltk.corpus import stopwords
import nltk
nltk.download("stopwords")

```

```python
stopwords.words('english')
```

```python
len(stopwords.words('english'))
```

```python
def remove_stopwords(text):
  new_text = []

  for word in text.split():
    if word in stopwords.words('english'):
      new_text.append('')
    else:
      new_text.append(word)
  x = new_text[:]
  new_text.clear()
  return " ".join(x)
```

```python
df['review'] = df['review'].apply(remove_stopwords)
```

#remove_emoji handle

```python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"
                           u"\U0001F300-\U0001F5FF"
                           u"\U0001F680-\U0001F6FF"
                           u"\U0001F1E0-\U0001F1FF"
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji("game is on 🔥🔥")
```

#Stemmer


```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
  return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = "walk walks walking walked"
stem_words(sample)
```

```python
#stemmer problem  meaning change for human readablity
text = "probably my alltime favorite movie a story of selflessness sacrifice and dedication to a noble cause but its not preachy or boring it just never gets old despite my having seen it some 15 or more times in the last 25 years paul lukas performance brings tears to my eyes and bette"
stem_words(text)
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
wordnet_lematizer = WordNetLemmatizer()


sentense = "He was running and eating at same time. He has bad habit of swimming after playing long"
punctuation="?:!.,;"
sentence_words = nltk.word_tokenize(sentense)
for word in sentence_words:
  if word in punctuation:
    sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
  print("{0:20}{1:20}".format(word,wordnet_lematizer.lemmatize(word , pos='v')))



```
